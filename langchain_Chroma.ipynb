```python
from langchain_openai import OpenAIEmbeddings
from langchain.vectorstores import Chroma
```

```python
from langchain.schema import Document

# Create LangChain documents for famous physicists

document_1 = Document(
    page_content="Albert Einstein revolutionized modern physics with his theory of relativity. His work laid the foundation for understanding space, time, and gravity, and he received the Nobel Prize for his explanation of the photoelectric effect.",
    metadata={"field": "Relativity"}
)

document_2 = Document(
    page_content="Isaac Newton formulated the laws of motion and universal gravitation. His work in classical mechanics remained the cornerstone of physics for over two centuries.",
    metadata={"field": "Classical Mechanics"}
)

document_3 = Document(
    page_content="Marie Curie conducted pioneering research on radioactivity, becoming the first person to win two Nobel Prizes in different scientific fields—Physics and Chemistry.",
    metadata={"field": "Radioactivity"}
)

document_4 = Document(
    page_content="Richard Feynman was known for his work in quantum electrodynamics and his unique teaching style. He made complex physics concepts accessible and won the Nobel Prize in 1965.",
    metadata={"field": "Quantum Electrodynamics"}
)

document_5 = Document(
    page_content="Stephen Hawking made significant contributions to cosmology and black hole physics. His theoretical prediction that black holes emit radiation, now known as Hawking radiation, changed the understanding of black holes.",
    metadata={"field": "Cosmology"}
)
```

```python
all_docs = [document_1, document_2, document_3, document_4, document_5]
```

```python
all_docs
```

```python
vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory='chroma_db',
    collection_name='sample'
)
```

```python
#Adding Documents
vector_store.add_documents(all_docs)
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
# search documents
vector_store.similarity_search(
    query='Who among them are related to modern physics?',
    k=1
)
```

```python
# search with similarity score
vector_store.similarity_search_with_score(
    query='Who among them are related to modern physics?',
    k=2
)
```

```python
# meta-data filter
vector_store.similarity_search_with_score(
    query="",
    filter={"field": "Radioactivity"}
)
```
